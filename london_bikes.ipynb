```python
#Specifying the column names
new_cols_dict = {
    'timestamp':'time',
    'cnt':'count',
    't1':'temp_real_C',
    't2':'temp_feels_like_C',
    'hum':'humidity_percent',
    'wind_speed': 'wind_speed_kph',
    'weather_code':'weather',
    'is_holiday':'is_holiday',
    'is_weekend':'is_weekend',
    'season':'season'
}

#Renaming the columns
df.rename(new_cols_dict, axis=1, inplace=True)
```

```python
#Changing humidity values to percentage
df.humidity_percent = df.humidity_percent / 100
```

```python
#Creating dictionary to map integers to written values in season column
season_dict = {
    '0.0':'spring',
    '1.0':'summer',
    '2.0':'autumn',
    '3.0':'winter'
}

#Creating dictionary to map integers to written values in weather column
weather_dict = {
    '1.0':'Clear',
    '2.0':'Scattered clouds',
    '3.0':'Broken clouds',
    '4.0':'Cloudy',
    '7.0':'Rain',
    '10.0':'Rain with thunderstorm',
    '26.0':'Snowfall'
}
```

```python
#Changing integers to strings
df.season = df.season.astype('str')
df.weather = df.weather.astype('str')

#Mapping values to written values
df.season = df.season.map(season_dict)
df.weather = df.weather.map(weather_dict)
```
